```python
from pathlib import Path
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Notebook is expected inside the notebooks folder.
PROJECT_DIR = Path.cwd().parent if Path.cwd().name.lower() == "notebooks" else Path.cwd()
RAW_DIR = PROJECT_DIR / "data" / "raw"
PROCESSED_DIR = PROJECT_DIR / "data" / "processed"
REPORT_DIR = PROJECT_DIR / "reports"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
REPORT_DIR.mkdir(parents=True, exist_ok=True)

print("Project:", PROJECT_DIR)
print("Raw data:", RAW_DIR)
```

```python
csv_files = {
    "customers": "olist_customers_dataset.csv",
    "geolocation": "olist_geolocation_dataset.csv",
    "orders": "olist_orders_dataset.csv",
    "order_items": "olist_order_items_dataset.csv",
    "payments": "olist_order_payments_dataset.csv",
    "reviews": "olist_order_reviews_dataset.csv",
    "products": "olist_products_dataset.csv",
    "sellers": "olist_sellers_dataset.csv",
    "category_translation": "product_category_name_translation.csv",
}

data = {}
for name, filename in csv_files.items():
    file_path = RAW_DIR / filename
    if not file_path.exists():
        raise FileNotFoundError(f"Missing file: {file_path}. Check data/raw/")
    data[name] = pd.read_csv(file_path)
    print(f"{name:22s} {data[name].shape[0]:,} rows x {data[name].shape[1]} columns")

print("All nine files loaded.")
```

```python
for name, df in data.items():
    print("\\nTABLE:", name)
    print("Shape:", df.shape)
    print("Exact duplicate rows:", df.duplicated().sum())
    display(df.isna().sum().to_frame("missing_count").query("missing_count > 0"))
```

```python
for name, df in data.items():
    before = len(df)
    df = df.drop_duplicates().copy()
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].map(lambda x: x.strip() if isinstance(x, str) else x)
    data[name] = df
    print(f"{name}: removed {before - len(df):,} exact duplicate rows")
```

```python
date_map = {
    "orders": [
        "order_purchase_timestamp", "order_approved_at",
        "order_delivered_carrier_date", "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ],
    "reviews": ["review_creation_date", "review_answer_timestamp"]
}
for table, columns in date_map.items():
    for col in columns:
        if col in data[table].columns:
            data[table][col] = pd.to_datetime(data[table][col], errors="coerce")

display(data["orders"].dtypes.to_frame("data_type"))
```

```python
orders = data["orders"]
products = data["products"]

for col in ["order_approved_at", "order_delivered_carrier_date",
            "order_delivered_customer_date"]:
    if col in orders.columns:
        orders[col + "_was_missing"] = orders[col].isna()

product_numeric = [
    "product_name_lenght", "product_description_lenght",
    "product_photos_qty", "product_weight_g", "product_length_cm",
    "product_height_cm", "product_width_cm"
]
for col in product_numeric:
    if col in products.columns:
        products[col + "_was_missing"] = products[col].isna()
        median = products[col].median()
        if pd.notna(median):
            products[col] = products[col].fillna(median)

if "product_category_name" in products.columns:
    products["product_category_name"] = products["product_category_name"].fillna("unknown")

data["orders"] = orders
data["products"] = products
print("Missing-value handling completed.")
```

```python
orders["actual_delivery_days"] = (
    orders["order_delivered_customer_date"] - orders["order_purchase_timestamp"]
).dt.total_seconds() / 86400

orders["estimated_delivery_days"] = (
    orders["order_estimated_delivery_date"] - orders["order_purchase_timestamp"]
).dt.total_seconds() / 86400

orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"] - orders["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

orders["delivery_result"] = np.select(
    [orders["order_delivered_customer_date"].isna(),
     orders["delivery_delay_days"] > 0],
    ["Unknown", "Late"],
    default="On time"
)
data["orders"] = orders

display(orders["delivery_result"].value_counts(dropna=False).to_frame("orders"))
display(orders["actual_delivery_days"].describe().round(2).to_frame())
```

```python
outlier_rows = []
for col in ["actual_delivery_days", "delivery_delay_days"]:
    values = orders[col].dropna()
    if len(values):
        q1, q3 = values.quantile([0.25, 0.75])
        iqr = q3 - q1
        lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
        flag = col + "_outlier_flag"
        orders[flag] = (orders[col] < lower) | (orders[col] > upper)
        outlier_rows.append({
            "column": col, "lower_bound": lower,
            "upper_bound": upper, "outliers_flagged": int(orders[flag].sum())
        })
outlier_summary = pd.DataFrame(outlier_rows)
display(outlier_summary)
```

```python
scale_cols = [c for c in [
    "actual_delivery_days", "estimated_delivery_days", "delivery_delay_days"
] if c in orders.columns]

valid = orders[scale_cols].notna().all(axis=1)
if valid.any():
    mm = MinMaxScaler()
    zs = StandardScaler()
    mm_cols = [c + "_minmax" for c in scale_cols]
    zs_cols = [c + "_zscore" for c in scale_cols]
    orders.loc[valid, mm_cols] = mm.fit_transform(orders.loc[valid, scale_cols])
    orders.loc[valid, zs_cols] = zs.fit_transform(orders.loc[valid, scale_cols])

data["orders"] = orders
display(orders.head())
```

```python
order_items = data["order_items"]
products = data["products"]
sellers = data["sellers"]

logistics_items = order_items.merge(
    orders, on="order_id", how="left", validate="many_to_one"
)

product_cols = [c for c in products.columns
                if c == "product_id" or c not in logistics_items.columns]
logistics_items = logistics_items.merge(
    products[product_cols], on="product_id",
    how="left", validate="many_to_one"
)

seller_cols = [c for c in sellers.columns
               if c == "seller_id" or c not in logistics_items.columns]
logistics_items = logistics_items.merge(
    sellers[seller_cols], on="seller_id",
    how="left", validate="many_to_one"
)

print("Joined table shape:", logistics_items.shape)
display(logistics_items.head())
```

```python
for name, df in data.items():
    df.to_csv(PROCESSED_DIR / f"{name}_cleaned.csv", index=False)

orders.to_csv(PROCESSED_DIR / "orders_preprocessed.csv", index=False)
logistics_items.to_csv(PROCESSED_DIR / "logistics_items_analysis.csv", index=False)

print("Processed files saved to:", PROCESSED_DIR)
```

```python
report = ["WEEK 2 DATA QUALITY SUMMARY", "=" * 55]

for name, df in data.items():
    report += [
        f"\\nTABLE: {name}",
        f"Rows: {len(df)}",
        f"Columns: {len(df.columns)}",
        f"Exact duplicate rows remaining: {df.duplicated().sum()}",
        "Missing values by column:",
        df.isna().sum().to_string()
    ]

report += ["\\nOUTLIER SUMMARY"]
report.append(outlier_summary.to_string(index=False) if len(outlier_summary)
              else "No eligible delivery fields.")
report += ["\\nDELIVERY RESULT COUNTS",
           orders["delivery_result"].value_counts(dropna=False).to_string(),
           "\\nOriginal data/raw files were not modified."]

report_path = REPORT_DIR / "week2_data_quality_summary.txt"
report_path.write_text("\\n".join(report), encoding="utf-8")
print("Audit report saved:", report_path)
print("WEEK 2 PREPROCESSING COMPLETED")
```
